```python
from sklearn.model_selection import train_test_split,StratifiedKFold
import pandas as pd
import numpy as np
from sys import argv
import pandas as pd

random_seed=42
```

```python
if len (argv) < 3 :
  print("error")
  exit()
else:
  d1 = argv[1]
  d2 = argv[2]
```

```python
d1 = pd.read_csv('/home/um40/LB2_project_Group_1/Data_preparation/Data_cleaning/negative_set.clean', sep='\t')
d2 = pd.read_csv('/home/um40/LB2_project_Group_1/Data_preparation/Data_cleaning/positive_set.clean', sep='\t')
```

assegno ad ogni colona un nome e concateno i df pos e neg

```python
column_names = ["class", "ID", "organism_name", "kingdom", "length", "cleavage_pos", "early_trans"]


positive = pd.read_csv(d1, sep="\t", header = None, names = column_names)
negative = pd.read_csv(d2, sep="\t", header = None, names = column_names)

df = pd.concat([positive, negative], ignore_index=True)
```

```python
y = df['class']
X = df.drop(['class'], axis=1)
```

# **Splitting into train(0.8) and test(0.2)**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=random_seed
    )
```

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=random_seed)
```

creo colonna fold e dopo nel for assegno ad ogni record il fold corrispondente

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=random_seed)
df['fold'] = -1

for fold_idx, (train_index, val_index) in enumerate(skf.split(X_train, y_train), start=1):

    X_fold_train, X_fold_val = X_train.iloc[train_index], X_train.iloc[val_index]
    y_fold_train, y_fold_val = y_train.iloc[train_index], y_train.iloc[val_index]

    actual_val_indices = X_train.index[val_index]

    df.loc[actual_val_indices, 'fold'] = fold_idx
```

salvo tutti i file

```python
df.to_csv('Cleaned_merged_dataset.tsv', sep="\t")
X_train.to_csv('X_train.tsv', sep="\t")
X_test.to_csv('X_test.tsv', sep="\t")
y_train.to_csv('y_train.tsv', sep="\t")
y_test.to_csv('y_test.tsv', sep="\t")
```
