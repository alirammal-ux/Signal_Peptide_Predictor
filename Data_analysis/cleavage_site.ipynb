# Data analysis - sequence logo of cleavage site

## 1. Extract cleavage region for training and test set

From SP-containing sequences (= positive set), align the cleavage-site region [-13, +2] and build a logo.


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv('../Data_preparation/Cleaned_merged_dataset.tsv', sep='\t')
data
```

```python
fasta_file = '../Data_preparation/Data_cleaning/positive_set.cluster/cluster-results-positive_rep_seq.fasta'
```

Open the fasta file containing all sequences of the representatives in the positive set. Then extract the region [-13,+2] around the cleavage site and save separately if that seq belongs to training set or test set (respectivelly cleavage_site_train and cleavage_site_test).

```python
cleavage_site_train=list()
cleavage_site_test=list()

with open(fasta_file, 'r') as fasta:
    for line in fasta:
        if line.startswith('>'):
            UniprotID = line.rstrip()[1:]
            fold = data.loc[data['ID'] == UniprotID, 'fold'].values[0] # estraggo la riga corrispondente ad UniprotID e ricavo il valore di fold per determinare a che set appartiene
            cleavage_pos = data.loc[data['ID'] == UniprotID, 'cleavage_pos'].values[0]
            start_pos = cleavage_pos - 13
            end_pos = cleavage_pos + 2
            seq = fasta.readline()

            if fold == -1: 
                # Test set
                cleavage_site_test.append(seq[start_pos:end_pos]) # keep just the region [-13,+2] around the cleavage site
            else:
                # Training set
                cleavage_site_train.append(seq[start_pos:end_pos]) # keep just the region [-13,+2] around the cleavage site
```

Save all the sequences of cleavage regions in external txt files, respectivelly 'cleavage_region_train.txt' for training set seq and 'cleavage_region_test.txt' for test set seq

```python
with open('cleavage_region_train.txt', 'w') as txt:
    for seq in cleavage_site_train:
        print(seq, file=txt)
```

```python
with open('cleavage_region_test.txt', 'w') as txt:
    for seq in cleavage_site_test:
        print(seq, file=txt)
```

## 4. Analysis of the results

As we can see both sequence logo of training and test sets are similar (this means that there is no internal bias in the two set).

By looking to the single logo, we can easily identify some expected structure:
- the first 7 positions of the logo (namely the -13:-6 pos with respect cleavage site in original seq) are dominates by leucine residues. Due to its hydrophobic property, we expect this result from our knowledge of the hydrophobic region in the signal peptide.
- position 9, 10 of the logo is not informative (no predominant residue)
- in region [13, 15] of the logo we can identify the typical AXA motif (unfurtunately too short to be used as marker)
