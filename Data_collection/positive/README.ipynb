We started collecting the preliminary dataset as two set: positive data and negative data.

In positive data we filtered proteins with *signal peptides* directly from UniprotKB using the following fields:
- Signal peptide only with experimental evidence
- Reviewed
- No fragments
- Eukaryotic proteins (Taxonomy: 2759)
- Sequence length >= 40 residues
- Existence of protein at protein level

and we downloaded the [URL](https://rest.uniprot.org/uniprotkb/search?compressed=false&format=json&query=%28%28ft_signal_exp%3A*%29+AND+%28reviewed%3Atrue%29+AND+%28fragment%3Afalse%29+AND+%28taxonomy_id%3A2759%29+AND+%28length%3A%5B40+TO+*%5D%29+AND+%28existence%3A1%29%29&size=500) for API, this first is called <u>preliminary positive dataset</u> from which we filtered for the others characteristics, specifically:

We have created a function called *filter_function*, this function takes in input a single entry and returns True if the entry has a signal peptide that starts at the beginning of the chain and is longer or equal to 14 amino acids, excluding not cleaved ones, and return False otherwise.

After that we have created the function called *extract_fields*, this function takes in input a single entry and return some fields of it like: UniProtAccession, Organism name, Eukaryotic kingdom (Metazoa, Fungi, Plants, Other), Protein length and the position of the signal-peptide cleavage site.


In negative data we filtered proteins with *no signal peptides* directly from UniprotKB by some fields:
- Reviewed
- No fragments
- Eukaryotic proteins (Taxonomy: 2759)
- Sequence length >= 40 residues
- No signal peptides at any evidence at protein level
- Experimental located in: cytosol, nucleus, mitochondrion, plastid, perozisome, cell membrane

and we downloaded the [URL](https://rest.uniprot.org/uniprotkb/search?compressed=false&format=json&query=%28%28taxonomy_id%3A2759%29+AND+%28existence%3A1%29+AND+%28reviewed%3Atrue%29+AND+%28fragment%3Afalse%29+AND+%28length%3A%5B40+TO+*%5D%29+AND+%28%28cc_scl_term_exp%3ASL-0091%29+OR+%28cc_scl_term_exp%3ASL-0191%29+OR+%28cc_scl_term_exp%3ASL-0173%29+OR+%28cc_scl_term_exp%3ASL-0039%29+OR+%28cc_scl_term_exp%3ASL-0209%29+OR+%28cc_scl_term_exp%3ASL-0204%29%29+NOT+%28ft_signal%3A*%29%29) for API from which we have a <u>preliminary negative dataset</u> and from which we filtered for the others characteristics, specifically:

And with our function called *extract_fields* we obtained filtered for the subsequents fields and directly returns five columns:
- UniProt accession
- Organism name
- Eukaryotic kingdom (Metazoa, Fungi, Plants, Other)
- Protein length
- Whether a transmembrane helix starts within the first 90 residues (true / false)

----------------------------------------------------------------

In the end we have created the function *get_dataset*, it takes three parameters in the case of negative dataset: 
+ (STRING) The URL corresponding to the specific API call
+ (FUNCTION) A function to filter entries according to specific criteria and returns directly fields we are looking for
+ (STRING) The output file name where TSV will be saved
and four parameters in the case of positive dataset:
+ (STRING)The URL corresponding to the specific API call
+ (FUNCTION)A filter function to filter entries according to specific criteria
+ (FUNCTION)An extract function used to get some specific fields of interest from the JSON format.
+ (STRING)The output file name where TSV will be saved
